```python
invalid_mask = (
    (df["ap_hi"] < 70) | (df["ap_hi"] > 250) |
    (df["ap_lo"] < 40) | (df["ap_lo"] > 150) |
    (df["height"] < 120) | (df["height"] > 220) |
    (df["weight"] < 30) | (df["weight"] > 200)
)

print("Total invalid rows:", invalid_mask.sum())
```

```python
print("Original rows:", len(df))
print("Invalid rows:", invalid_mask.sum())
print("Rows after cleaning:", len(df) - invalid_mask.sum())
print("Percentage retained:", (1 - invalid_mask.mean()) * 100)
```

```python
df_clean = df[~invalid_mask].copy()

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("Invalid ap_hi:", ((df_clean["ap_hi"] < 70) | (df_clean["ap_hi"] > 250)).sum())
print("Invalid ap_lo:", ((df_clean["ap_lo"] < 40) | (df_clean["ap_lo"] > 150)).sum())
print("Invalid height:", ((df_clean["height"] < 120) | (df_clean["height"] > 220)).sum())
print("Invalid weight:", ((df_clean["weight"] < 30) | (df_clean["weight"] > 200)).sum())
```
