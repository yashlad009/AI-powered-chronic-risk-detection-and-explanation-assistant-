```python
import os
import numpy as np
import pandas as pd
from tensorflow import keras

print("Libraries imported successfully.")
```

```python
# Cell 2: Load the trained ANN model and scaler

import joblib

model_path = "../artifacts/cvd/tuned_ann.keras"
scaler_path = "../artifacts/cvd/scaler.pkl"

risk_model = keras.models.load_model(model_path)
scaler = joblib.load(scaler_path)

print("Tuned ANN model loaded successfully.")
print("Scaler loaded successfully.")
```

```python
# Cell 3: Create sample patient input

patient_data = pd.DataFrame([{
    "age": 50,
    "gender": 0,
    "height": 170,
    "weight": 70,
    "ap_hi": 120,
    "ap_lo": 80,
    "cholesterol": 1,
    "gluc": 1,
    "smoke": 0,
    "alco": 0,
    "active": 1
}])

print("Patient input:")
display(patient_data)
```

```python
# Cell 4: Inspect the saved scaler

print("Scaler type:", type(scaler).__name__)
print("Number of features expected:", scaler.n_features_in_)

if hasattr(scaler, "feature_names_in_"):
    print("Features used when scaler was fitted:")
    print(list(scaler.feature_names_in_))
else:
    print("Scaler does not contain feature names.")
```

```python
# Cell 5: Prepare patient input for the ANN model

numerical_features = [
    "age",
    "height",
    "weight",
    "ap_hi",
    "ap_lo"
]

patient_scaled = patient_data.copy()

# Scale only the numerical features
patient_scaled[numerical_features] = scaler.transform(
    patient_data[numerical_features]
)

print("Patient data after scaling:")
display(patient_scaled)
```

```python
# Cell 6: Make CVD risk prediction

prediction_probability = risk_model.predict(
    patient_scaled,
    verbose=0
)[0][0]

prediction_class = int(prediction_probability >= 0.5)

print("Predicted probability:", prediction_probability)
print("Predicted class:", prediction_class)

if prediction_class == 1:
    print("⚠️ Higher risk of cardiovascular disease")
else:
    print("✅ Lower risk of cardiovascular disease")
```
