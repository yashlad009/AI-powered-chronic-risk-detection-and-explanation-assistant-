```python
# ------------------------------Cleaning the data ----------------------------
```

```python
# we need to replace 0 in glucose , bp , skinthickness , insulin , BMI  , ----- pregnancy 0 is valid , age has no 0 so thats also valid 
```

```python
cols = [
    "Glucose" ,
    "BloodPressure" ,
    "SkinThickness" ,
    "Insulin",
    "BMI"
    ]
```

```python
df[cols] = df[cols].replace(0,np.nan)
```

```python
(df == 0).sum()
```

```python
df.isnull().sum()
```

```python
# now check for outliers 
# if outliers are present in column then np.NAN replaced with its median , if not present then with mean
# Check the distribution of each feature.
# If extreme outliers significantly affect the data, use the median.
# Otherwise, use the mean to impute missing values.
import matplotlib.pyplot as plt

df.boxplot(column=["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"])
plt.xticks(rotation=45)
plt.show()
```

```python
# Glucose , blood pressure and BMI has no significant outliers so we replace nan by its mean

df["Glucose"] = df["Glucose"].fillna(
    df["Glucose"].mean()
)
 
df["BloodPressure"] = df["BloodPressure"].fillna(
    df["BloodPressure"].mean()
)

df["BMI"] = df["BMI"].fillna(
    df["BMI"].mean()
)
```

```python
# SkinThickness and Insulin has skewed ouliers so we replace its nan values by np.median

df["SkinThickness"] = df["SkinThickness"].fillna(
    df["SkinThickness"].median()
)
df["Insulin"] = df["Insulin"].fillna(
    df["Insulin"].median()
)
 
```

```python
# ok now see why need of feature scaling 

# now our features are like glucose , age , insulin 
# suppose a patient has 
#     glucose - 148
#     age - 50
#     insulin - 155.548

# all features are having its different ranges so we use neural network and initially is see only no so we use feature scaling
# so with the help of feature scaling we can set a same range for features 

# for that we have 2 methods 

# 1. StandardScaling method 
# 2.MinMax Scaling method

# we use standard scaling feature 
# in that set like 
# mean ~ 0
# s.d ~ 1

# formula 
#     z = (x - u ) / s.d  ->   x = original value   u = mean 


# for that we make a set x where all features where we can apply the feature scaling and in set y there is target
```
